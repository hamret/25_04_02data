```python
import requests

url = 'http://openapi.seoul.go.kr:8088/(메모장열어서개인코드집어넣을것)/json/viewNightSpot/1/900/'

r = requests.get(url)
data = r.json()
data
```

```python
import pandas as pd
temp=[]
for d in data['viewNightSpot']['row']:
  temp.append(d)
  placedata = pd.DataFrame(temp)
placedata
```

```python
for n, row in placedata.iterrows():
  print(n, row['TITLE'], row['LA'], row['LO'])
```

```python
import folium

folium.Map()
```

```python
import folium

folium.Map(location= [37.4965, 126.9573],
           zoom_start=17)
```

```python
plmap = folium.Map(location=[37.538240218883, 127.09689071008],
                   zoom_start=15,
                   zoom_control=False,
                   control_scale= True)
plmap
```

```python

for n, row in placedata.iterrows():
  vm = folium.Marker(location=[row['LA'],row['LO']], popup = folium.Popup(row['TITLE'], max_width=200))
  vm.add_to(plmap)

plmap
```

```python
plmap2 = folium.Map(location=[37.538240218883, 127.09689071008],
                   zoom_start=15,
                   zoom_control=False,
                   control_scale= True)
plmap2
```

```python
from folium.plugins import MarkerCluster

plVar = MarkerCluster().add_to(plmap2)

for n, row in placedata.iterrows():
  folium.Marker(location=[row['LA'],row['LO']], popup = folium.Popup(row['TITLE'], max_width=200)).add_to(plVar)

plmap2
```
